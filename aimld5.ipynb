Detect outliers using statistical methods (IQR or Z-score) and visualize them
using box plots.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Identify numerical columns
numerical_cols = df_cleaned.select_dtypes(include=['int64', 'float64']).columns

print("Numerical columns identified for outlier detection:", list(numerical_cols))
```

```python
# Function to detect outliers using IQR
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers

# Detect and report outliers for each numerical column
print("\nOutlier Detection Results (IQR Method):")
for col in numerical_cols:
    outliers = detect_outliers_iqr(df_cleaned, col)
    if not outliers.empty:
        print(f"  - Column '{col}': {len(outliers)} outliers detected.")
    else:
        print(f"  - Column '{col}': No outliers detected.")
```

```python
# Visualize outliers using box plots for each numerical column
plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(3, 3, i + 1) # Adjust subplot grid as needed
    sns.boxplot(y=df_cleaned[col])
    plt.title(f'Box Plot of {col}')
    plt.ylabel('')
plt.tight_layout()
plt.show()
```

 Handle outliers using suitable preprocessing techniques.

```python
# Create a copy of df_cleaned to avoid modifying the original DataFrame
df_outliers_handled = df_cleaned.copy()

# Function to cap outliers using IQR
def cap_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)
    return df

# Apply capping to each numerical column where outliers were detected
# Student_ID, Tool_Diversity, Anxiety_Level_During_Exams had no outliers, so we don't need to cap them
columns_to_cap = [
    'Pre_Semester_GPA',
    'Weekly_GenAI_Hours',
    'Traditional_Study_Hours',
    'Perceived_AI_Dependency',
    'Post_Semester_GPA',
    'Skill_Retention_Score'
]

for col in columns_to_cap:
    df_outliers_handled = cap_outliers_iqr(df_outliers_handled, col)

print("Outliers have been capped in the specified numerical columns.")
print("First 5 rows of the DataFrame after outlier handling:")
display(df_outliers_handled.head())
```

```python
# Visualize outliers again using box plots for each numerical column after capping
plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(3, 3, i + 1) # Adjust subplot grid as needed
    sns.boxplot(y=df_outliers_handled[col]) # Use df_outliers_handled
    plt.title(f'Box Plot of {col} (After Capping)')
    plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import pandas as pd

# Create a copy of df_outliers_handled to perform encoding
df_encoded = df_outliers_handled.copy()

# Identify categorical columns
categorical_cols = df_encoded.select_dtypes(include='object').columns
print(f"Categorical columns identified: {list(categorical_cols)}")

# --- Label Encoding ---
# Columns with a natural order (ordinal)
# 'Year_of_Study': Freshman < Sophomore < Junior < Senior
# 'Burnout_Risk_Level': Low < Medium < High

# Initialize LabelEncoder
le = LabelEncoder()

# Apply Label Encoding to 'Year_of_Study'
# Define the order for Year_of_Study
year_of_study_order = ['Freshman', 'Sophomore', 'Junior', 'Senior']

# Check for any values in 'Year_of_Study' that are not in year_of_study_order
missing_year_values = set(df_encoded['Year_of_Study']) - set(year_of_study_order)
if missing_year_values:
    print(f"Warning: The following values in 'Year_of_Study' were not in the defined order: {missing_year_values}. Adding them to the order.")
    # Dynamically add any missing values to the end of the order, maintaining original order for existing ones
    for val in missing_year_values:
        year_of_study_order.append(val)

mapping_year_of_study = {category: i for i, category in enumerate(year_of_study_order)}
df_encoded['Year_of_Study_Encoded'] = df_encoded['Year_of_Study'].map(mapping_year_of_study)
print("Label Encoded 'Year_of_Study' column created.")

# Apply Label Encoding to 'Burnout_Risk_Level'
# Define the order for Burnout_Risk_Level
burnout_risk_order = ['Low', 'Medium', 'High']
mapping_burnout_risk = {category: i for i, category in enumerate(burnout_risk_order)}
df_encoded['Burnout_Risk_Level_Encoded'] = df_encoded['Burnout_Risk_Level'].map(mapping_burnout_risk)
print("Label Encoded 'Burnout_Risk_Level' column created.")

# Drop original Label Encoded columns
df_encoded = df_encoded.drop(columns=['Year_of_Study', 'Burnout_Risk_Level'])

# --- One-Hot Encoding ---
# Columns without a natural order (nominal)
# 'Major_Category', 'Primary_Use_Case', 'Prompt_Engineering_Skill', 'Institutional_Policy'

# Identify columns for One-Hot Encoding
one_hot_cols = ['Major_Category', 'Primary_Use_Case', 'Prompt_Engineering_Skill', 'Institutional_Policy']

# Apply One-Hot Encoding
df_encoded = pd.get_dummies(df_encoded, columns=one_hot_cols, drop_first=True, dtype=int)
print("One-Hot Encoded columns created.")

print("\nFirst 5 rows of the DataFrame after encoding:")
display(df_encoded.head())

print("\nDataFrame Info after encoding:")
df_encoded.info()
```

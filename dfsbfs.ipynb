```python
import pandas as pd

# Load the uploaded dataset
df = pd.read_csv('/content/graph_dataset (1).csv')
display(df.head())
```

```python
from collections import defaultdict

# Build the graph dynamically from the dataset
# Assuming the CSV contains 'source' and 'target' or similar edge columns
graph = defaultdict(list)

# Identify column names automatically
source_col = df.columns[0]
target_col = df.columns[1]

for _, row in df.iterrows():
    u = str(row[source_col])
    v = str(row[target_col])
    # Assuming an undirected graph; modify if directed
    graph[u].append(v)
    graph[v].append(u)

# Convert defaultdict back to normal dict for display
graph = dict(graph)
print(f"Dynamically constructed graph with {len(graph)} nodes.")
```

                  DFS

```python
def dfs(graph, start_node, visited=None, traversal=None):
    if visited is None:
        visited = set()
    if traversal is None:
        traversal = []

    if start_node not in graph:
        return f"Node '{start_node}' not found in the graph."

    visited.add(start_node)
    traversal.append(start_node)

    # Recursively visit all unvisited neighbors
    for neighbor in graph[start_node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited, traversal)

    return traversal

# Automatically pick the first node in your graph as the starting point
start_node = list(graph.keys())[0]

print(f"DFS Traversal starting from '{start_node}':")
dfs_result = dfs(graph, start_node)
print(" -> ".join(dfs_result))
```
